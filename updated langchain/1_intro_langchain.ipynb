```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GOOGLE_API_KEY"]= os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]= os.getenv("GROQ_API_KEY")
```

```python
from langchain.agents import create_agent

def get_weather(city:str)->str:
    "gets the weather of the city"
    return f"the weather of the city {city} is very rainy"

agent = create_agent(
    model = "google_genai:gemini-3-flash-preview",
    tools=[get_weather],
    system_prompt = "you are a helpful assistant")

agent
```

```python
agent.invoke({"messages":[{"role": "user","content":"give me weather of newyork" }]})
```

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

model = init_chat_model("google_genai:gemini-3-flash-preview")

response = model.invoke("Why do parrots talk?")
print(response.content[-1])
```
